# 이동평균

(시계열분석 및 계절성의 근거 마련 가능)

이동평균은 시계열에서 연속된 몇개의 값을 묶어서 평균을 내고, 그 묶음을 한칸씩 옮기며 반복해 만드는 새로운 시계열을 의미.

ex) 파도가 치는 바닷가에서 물높이를 재면 파도 각각에 따라 값이 요동친다.
    but, 최근 몇번의 측정값을 평균내면 요동치는 값을 상쇄되고, 밀물 썰물과 같은 큰 흐름만 남는다.

이동평균은 가파른 움직임을 걸러내고 느린 흐름만 남기는 필터라고 할수있다.

- k개월 후행 이동평균 = MA_k(t) = (y(t)+y(t-1)+...+y(t-k+1)) / k
